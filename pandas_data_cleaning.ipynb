```python
import pandas as pd
import numpy as np
# Загрузка исходного датасета
acp = pd.read_csv('advertising_campaign_performance.csv')
```

```python
# Проверка типов данных и наличия пропусков
print(acp.info())
# Описательная статистика для выявления выбросов
print(acp.describe())
```

```python
# Защита от деления на ноль: замена 0 конверсий на NaN
acp['Conversion_safe'] = acp['Conversions'].replace(0, np.nan)
# Расчет CPA
acp['CPA'] = acp['Spend']/acp['Conversion_safe']
# Удаление промежуточной колонки
acp = acp.drop(columns = ['Conversion_safe'])
# Заполнение пропусков в CPA значениями из Spend
acp['CPA'] = acp['CPA'].fillna(acp['Spend'])
```

```python
# Проверка, что все пропуски устранены
print(acp.info())
```

```python
#Сохранение в файл
acp.to_csv('acp_clean.csv', index=False, encoding='utf-8-sig')
```
